# 4. Background and Literature

There are several studies related to this, but they often only focus on 1 or 2 socioeconomic factors. For example, we have a study from Sariaslan that studies how childhood family income affects violent criminality and substance misuse in Sweden. They found that children in the lowest income quintile had a 7 fold increase in criminality compared to peers in the highest quintile. Though this study does do analysis, it does not do spatial analysis. We do expect income will have a big factor in determining crime in a specific block. One study that does integrate spatial analysis in a particular city is from the University of Minnesota where they study socioeconomic status and demographic variables to crime in Pittsburgh. Here, they analyze census tract data from 1990 to 2001. They utilize multiple spatial analysis techniques through EDA. They used choropleth mapping to visualize crime across census tracts. Additionally, they used hotspot analysis to identify clusters of high and low crime areas. They also had to perform boundary reconciliation since they used two different years of census that had different boundaries. In terms of a model, they used a stepwise multiple linear regression. Using just socioeconomic factors, they were able to explain 21.1% of the variation in violent crime, with unemployment and household income as key predictors.  For our project, we will perform similar steps for the EDA to find potentially important features. We will use multi-step linear regression, but also other complex  models, such as kernel density estimation, to see whether they will result in lower loss. We will also include more socioeconomic and demographic variables, as we want to explain more of the variation.
