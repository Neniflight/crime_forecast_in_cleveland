# Background and Literature
**PART 4**

&emsp; Many studies that we looked at only focus on one or a few socioeconomic factors at a time. [A study from Saint Mary's University of Minnesota](https://gis.smumn.edu/GradProjects/MitchellS.pdf) performs an analysis that is very similar to what we planned. Using a variety of socioeconomic variables, they performed regression to predict hotspots and areas of high crime rates across Census tracts in Pittsburgh, Pennsylvania, and confirms a few already known facts about crime activity. Here, they analyze census tract data from 1990 to 2001. They utilize multiple spatial analysis techniques through EDA. They used choropleth mapping to visualize crime across census tracts. Additionally, they used hotspot analysis to identify clusters of high and low crime areas. They also had to perform boundary reconciliation since they used two different years of census that had different boundaries. In terms of a model, they used a stepwise multiple linear regression. Using just socioeconomic factors, they were able to explain 21.1% of the variation in violent crime, with unemployment rates and median household income as key predictors. In addition, the percentage of males aged 14-29 years showed correlation with rates of property crime.
<br></br>

They found that children in the lowest income quintile had a 7 fold increase in criminality later in life as compared to peers in the highest quintile.

## Census Data / ACS Code

```python
def download_OH_data(var_map, year_start, year_end):
    df_final = None
    for yr in range(year_start, year_end + 1):
        c = Census("4977648d549eae5dd6bc0563b7c148db6c44642d", year=yr)
        for raw_var, alias in var_map.items():
            if raw_var.startswith('DP'):
                data = c.acs5dp.get(
                    ('NAME', raw_var),
                    {'for': 'tract:*', 'in': f'state:{states.OH.fips} county:035'}
                )
            else:
                data = c.acs5.get(
                    ('NAME', raw_var),
                    {'for': 'tract:*', 'in': f'state:{states.OH.fips} county:035'}
                )
            df_temp = pd.DataFrame(data)
            df_temp.rename(columns={raw_var: f"{alias}_{yr}"}, inplace=True)
            # Merge it into df_final
            if df_final is None:
                # If this is the first chunk of data, just assign
                df_final = df_temp
            else:
                # Otherwise, merge on the geo-id columns
                df_final = pd.merge(
                    df_final, df_temp,
                    on=['NAME', 'state', 'county', 'tract'],
                    how='outer'  # or 'inner', your choice
                )             
    return df_final
```

# Analysis
**PART 9**

&emsp; We start our analysis our two datasets. The first step is to download all the codes from the years we are looking at using the Census API. We then collect the crime incidents from the City of Cleveland’s open data portal and filter out out only violent crimes (homicides and assaults) between the years 2015 and 2023. We focused on these crimes specifically because of the conclusions we came to in the literature, in addition to data contraint reasoning. After ETL, we combine these two datasets and place them into a GeoDataFrame. We ensure that there are no null/missing values in the dataframe, and filter out rows with tract_ids according to the data cleaning step. The pickle file that we are using to filter out tracts was generated in the "2010_to_2020_tracts.ipynb" notebook.
<br></br>
&emsp; We performed our spatial analysis on these datasets seperately. For the crime data, we used tools in ArcGIS to perform hot spot analyses on the filtered crime data. Showing us the areas of high and low crime zones across Cuyahoga county. This step gave us a sense of where violent crime was concentrated. We also ran a series of similar hot spot analyses on selected socioeconomic variables such as poverty indices, proportions of healthcare insured populations, median age, high school enrollment, and married-couple households—to identify which census tracts were significantly high or low in each category. Comparing these hotspot maps side by side suggested that poverty and certain family structure variables overlapped strongly with the areas of high violent crime, a pattern that is widely supported by related studies on socioeconomic disadvantage and its influence on criminal behavior. We also noticed moderate visual relationships in other variables like median age and high school enrollment. These overlays gave us a indication that some of the socioeconomic factors we picked may be meaningful predictors of crime variability across Cleveland tracts.

<br></br>
&emsp; We confirmed these observations by conducting a correlation analyses on the final merged dataset. Our results showed us some interesting findings. Firstly was that the 'middle-tier' income factors were actually negatively correlated with crime counts, while areas of high income or very little income had positive correlation. This somewhat makes sense and alligns with some of the literature we looked at, which suggested that areas of high income disparity, or areas where people grew up in poverty, were significantly more likely to see increased rates of crime. Access to healthcare was another negatively correlated factor. Perhaps the most interesting was the types of households. We found that the living alone were among the lowest predictors of crime. Female head households actually had a positive correlation with crime rates, while the male and married head households had a negative correlation with crime. with our literature review: multiple studies suggested that income disparities, unemployment, and measures of social disadvantage are key predictors. These correlations coincided with what we saw when analyzing our hotspots.

<br></br>
&emsp; Our final step was to perform predictive modeling to explore how well these socioeconomic variables explain crime levels. We used four different models in scikit-learn. First splitting our data into training and testing sets (80/20) then trained several models. These included k-Nearest Neighbors, Random Forest, Isolation Forest, and a one-class SVM approach. The kNN model had a very high correlation with an R^2 of .65, while the other two forest model have lower R^2 scores of 0.63. The mean squared error was also lower for the kNN model as well. The primary takeaway is that a well-tuned kNN. consistently performed best. These findings are different to prior work we looked at in the literature review, which found that found tree-based models particularly effective when dealing with socioeconomic variables. Overall, our analysis indicates that high rates of poverty, and areas of wide income distribution, are significant factors when determining the rates of crime in an area. Additionally high educational attainment and the type of household are also pretty good predictors of the number of crimes that occur in an area.

## Machine Learning

```python
y = final_df['crime_count']
X = final_df.iloc[:, 6:-2]
X
```

```python
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
X_scaled
```

```python
# 80/20 split for train/test
X_train, X_test, y_train, y_test = train_test_split(
    X_scaled, y, test_size=0.2, random_state=42)
```

### KNN

```python
param_grid = {"n_neighbors": range(1, 21)}
grid_search = GridSearchCV(KNeighborsRegressor(), param_grid, cv=5, scoring="neg_mean_squared_error")
grid_search.fit(X_train, y_train)
```

```python
best_k = grid_search.best_params_["n_neighbors"]
knn = KNeighborsRegressor(n_neighbors=best_k)
knn.fit(X_train, y_train)
```

```python
y_pred = knn.predict(X_test)
```

```python
mse_knn = mean_squared_error(y_test, y_pred)
r2_knn = r2_score(y_test, y_pred)
```

```python
print(f"Best k: {best_k}")
print(f"MSE: {mse_knn}")
print(f"R²: {r2_knn}")
```

### Random Forest

```python
forest_param_grid = {
    'n_estimators': [50, 100, 200],
    'max_depth': [None, 5, 10, 20],
    'min_samples_split': [2, 5, 10],
    'min_samples_leaf': [1, 2, 4]
}

rf = RandomForestRegressor(random_state=462)

grid_search = GridSearchCV(rf, forest_param_grid, scoring='neg_mean_squared_error', cv=5, n_jobs=-1)
grid_search.fit(X_train, y_train)

best_rf = grid_search.best_estimator_
```

```python
y_pred_rf = best_rf.predict(X_test)

mse_rf = mean_squared_error(y_test, y_pred_rf)
r2_rf = r2_score(y_test, y_pred_rf)
```

```python
print("Best Params:", grid_search.best_params_)
print("MSE:", mse_rf)
print("R²:", r2_rf)
```

### Isolation Forest

```python
iso = IsolationForest(contamination=0.05, random_state=314)

iso.fit(X_train)

inliers = iso.predict(X_train) == 1
X_train_in = X_train[inliers]
y_train_in = y_train[inliers]


rf = RandomForestRegressor(random_state=42)

grid_search = GridSearchCV(rf, forest_param_grid, scoring='neg_mean_squared_error', cv=5, n_jobs=-1)
grid_search.fit(X_train_in, y_train_in)
```

```python
y_pred = best_rf.predict(X_test)
mse_iso = mean_squared_error(y_test, y_pred)
r2_iso = r2_score(y_test, y_pred)
```

```python
print("Best Params:", grid_search.best_params_)
print("MSE:", mse_iso)
print("R²:", r2_iso)
```
